## Executive Summary: Optimal USA Gymnastics Lineup for 2024 Paris Olympics  

### Country of Recommendation  
This analysis focuses on selecting the optimal women's gymnastics lineup for Team USA at the 2024 Paris Olympics.  

### Goal and Optimality Criterion  
The objective is to maximize the total medal count for Team USA in women’s gymnastics. The optimization prioritizes gold medals, followed by silver and bronze, based on a simulation-driven approach.  

### Quantitative Forecast and Uncertainty  
The simulation results indicate that Team USA's expected total medal count is approximately **6.5 medals**, with a range from **3 to 12 medals**. The breakdown is as follows:  
- **Gold medals**: Expected **3**, with a range of **0 to 6**.  
- **Silver and bronze medals**: Expected **2 each**, with a range of **0 to 5**.  
The simulation incorporates random noise to account for real-world variability, reflecting uncertainty in individual and team performances.  

### Analysis Strategy  
1. **Data Collection**: Historical performance data was compiled from major competitions, including the **Tokyo 2021 Olympics**, **2022-2023 World Championships**, and other international events.  
2. **Performance Prediction**: A Linear Regression model was trained to estimate each gymnast’s score distribution based on historical results. Key features included competition recency and target-encoded average scores.  
3. **Simulation Framework**: 
   - Fixed lineups for **11 competing countries** based on historical performance.  
   - Predicted scores for all gymnasts, incorporating random noise for realism.  
   - Simulated all possible USA lineups, running **2000 simulations per lineup** to compute medal probabilities.  
4. **Results Interpretation**: The optimal lineup prioritized gold medals and included top gymnasts like **Simone Biles, Sunisa Lee, Jade Carey, and Jordan Chiles**. The final team selection aligned closely with real-world decisions, except for missing data on Hezly Rivera.  

The model provides a robust, data-driven approach to optimizing team selection and medal forecasting, with potential applications for future Olympic Games and other international competitions.
